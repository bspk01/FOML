```python
# -*- coding: utf-8 -*-

#All imports
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap
import random
import math
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml

########################################
#Part(a) #Synthetic data generation
########################################
#Define means and covariances
mean1 = [0, 0]
mean2 = [0, 1]
mean3 = [1, 0]
mean4 = [1, 1]
means = [mean1, mean2, mean3, mean4]
cov = [[0.1, 0], [0, 0.1]]  
data, labels = [], []   #intializing data and labels

#Sample data points from the bivariate Gaussian distribution
#You can use "np.random.multivariate_normal" function to sample the data points from the multivariate Gaussian distribution

for i in range(len(means)):
    samples = np.random.multivariate_normal(means[i], cov, 90)
    for sample in samples:
        data.append(sample)
    for label in [1 if i in [0, 3] else -1] * 90:
        labels.append(label)


#Generate training data

X_train = np.array([data[i * 90 + j] for i in range(4) for j in range(30)])
Y_train = np.array([labels[i * 90 + j] for i in range(4) for j in range(30)])

#Generate testing data

X_test = np.array([data[i * 90 + j] for i in range(4) for j in range(30, 90)])
Y_test = np.array([labels[i * 90 + j] for i in range(4) for j in range(30, 90)])

#Visualize the data using plt.scatter() function

plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.scatter(X_train[Y_train == 1][:, 0], X_train[Y_train == 1][:, 1],
            color='green', label='Class 1')
plt.scatter(X_train[Y_train == -1][:, 0], X_train[Y_train == -1][:, 1],
            color='blue', label='Class 2')
plt.title('Training Set')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()

plt.subplot(1, 2, 2)
plt.scatter(X_test[Y_test == 1][:, 0], X_test[Y_test == 1][:, 1],
            color='green', label='Class 1')
plt.scatter(X_test[Y_test == -1][:, 0], X_test[Y_test == -1][:, 1],
            color='blue', label='Class 2')
plt.title('Test Set')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()

plt.show()
########################################
#Part(b) #KNN classification
########################################

#Write "kNNClassify" function
class KNeighbourClassifier:
    """
    K-Nearest Neighbors Classifier
    This class implements the k-nearest neighbors algorithm for classification.
    """
    def __init__(self, n_neighbors=5,distance_metric="l2"):
        #Initializes the KNeighbourClassifier with the specified number of neighbors. Defaults to 5.
        #Default distance_metric is L2 norm

        self.n_neighbors = n_neighbors
        self.distance_metric = distance_metric
        self.X_train = None
        self.y_train = None

    def fit(self, X_train, Y_train):
        # This method fits the k-nearest neighbors classifier from the training data. X_train is the training data, represented as a numpy array of shape (n_samples, n_features), and Y_train is the target values of the training data, represented as a numpy array of shape (n_samples,)
        self.X_train = np.array(X_train, dtype=float)
        self.y_train = np.array(Y_train)

    def predict(self, X_test):
        # This method predicts the class labels for a set of data samples. X_test is the data to be predicted, represented as a numpy array of shape (n_samples, n_features)
        X_test = np.array(X_test, dtype=float)
        predictions = []
        for test_point in X_test:
            # Calculate distances from the test point to all the training points
            diff = self.X_train - test_point
            if self.distance_metric == "l2":
                distances = np.sqrt(np.sum(diff ** 2, axis=1))
            elif self.distance_metric == "l1":
                distances = np.sum(np.abs(diff), axis=1)
            elif self.distance_metric == "linf":
                distances = np.max(np.abs(diff), axis=1)
            else:
                raise ValueError("Unsupported distance metric: " + str(self.distance_metric))

            # Find k nearest neighbors
            nearest_neighbors = np.argsort(distances)[:self.n_neighbors]
            neighbor_labels = self.y_train[nearest_neighbors]
            
            #Assign the class of the majority of the k nearest neighbours
            values, counts = np.unique(neighbor_labels, return_counts=True)
            prediction = values[np.argmax(counts)]
            predictions.append(prediction)
        
        return np.array(predictions)


#Write "KNNAccuracy" function
def KNNAccuracy(true,pred):
    #Inputs : Ground truth and predicted labels
    #Outputs : Portion of data points that are correctly classified, i.e., accuracy
    true = np.asarray(true)
    pred = np.asarray(pred)
    return np.mean(true == pred)


#Create a visual representation of predictions
def visualize_predictions(X_test, Y_test, Y_pred):
    plt.figure(figsize=(8, 6))
    plt.scatter(X_test[Y_test == 1][:, 0], X_test[Y_test == 1][:, 1],
                color='green', label='Class 1')
    plt.scatter(X_test[Y_test == -1][:, 0], X_test[Y_test == -1][:, 1],
                color='blue', label='Class 2')
    plt.scatter(X_test[Y_pred != Y_test][:, 0], X_test[Y_pred != Y_test][:, 1],
                color='red', label='Misclassified')
    plt.title('Test Set Predictions')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.legend()
    plt.show()

#Generate and visualize the decision regions and overlay the test points

def visualize_decision_boundaries(classifier, X_train, Y_train, resolution=0.02):
    X_train = np.asarray(X_train)
    x_min, x_max = X_train[:, 0].min() - 1, X_train[:, 0].max() + 1
    y_min, y_max = X_train[:, 1].min() - 1, X_train[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, resolution),
                         np.arange(y_min, y_max, resolution))
    Z = np.array(classifier.predict(np.c_[xx.ravel(), yy.ravel()]))
    Z = Z.reshape(xx.shape)
    plt.contourf(xx, yy, Z, alpha=0.4, cmap=ListedColormap(('blue', 'green')))
    plt.scatter(X_train[:, 0], X_train[:, 1], c=Y_train, s=20, edgecolor='k')
    plt.title('Decision Boundaries')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.show()
   
knn = KNeighbourClassifier(n_neighbors=5)
knn.fit(X_train, Y_train)
predictions = knn.predict(X_test)

# Compute accuracy
accuracy = KNNAccuracy(Y_test, predictions)
print(f"KNN Classification Accuracy: {accuracy * 100:.4f}%")

visualize_predictions(X_test, Y_test, predictions)
visualize_decision_boundaries(knn, X_train, Y_train)   
    
####################################
#Part(c): Parameter selection: What is good value for k?
####################################
#Write holdoutCVkNN() Function
def holdoutCVkNN(k_range, rho, X_train, Y_train, num_rep=10):
  #Inputs: range of k values, fraction of training data to hold out, training data, training labels, number of repetitions
  #Outputs: average training and validation errors for every k in k_range
  #Iterate through range of k values
    tr_errors = np.zeros(len(k_range))
    val_errors = np.zeros(len(k_range))
    n_train = len(X_train)
    n_val = int(rho * n_train)
    for rep in range(num_rep):
        #Hold out rho fraction of training data in each repetition.
        indices = np.random.permutation(n_train)
        val_idx, tr_idx = indices[:n_val], indices[n_val:]
        X_tr, Y_tr = X_train[tr_idx], Y_train[tr_idx]
        X_val, Y_val = X_train[val_idx], Y_train[val_idx]
        for i, k in enumerate(k_range):
            #Train the KNN model on the remaining data and evaluate on the held out data
            knn_cv = KNeighbourClassifier(n_neighbors=k)
            knn_cv.fit(X_tr, Y_tr)
            tr_errors[i] += 1 - KNNAccuracy(Y_tr, knn_cv.predict(X_tr))
            val_errors[i] += 1 - KNNAccuracy(Y_val, knn_cv.predict(X_val))

  #Return errors on training and validation data (averaged over the repetitions)
    return tr_errors / num_rep, val_errors / num_rep

k_range = list(range(1, 22, 2))   #odd values of k to avoid ties
rho = 0.3
tr_errors, val_errors = holdoutCVkNN(k_range, rho, X_train, Y_train, num_rep=10)

#Plot training and validation errors for different values of k
plt.figure(figsize=(10, 6))
plt.plot(k_range, tr_errors, marker='o', label='Training error')
plt.plot(k_range, val_errors, marker='s', label='Validation error')
plt.xlabel('k (number of neighbours)')
plt.ylabel('Error')
plt.title('Hold-out cross validation (rho = 0.3, 10 repetitions)')
plt.xticks(k_range)
plt.legend()
plt.grid(True)
plt.show()

#what is the best value for k?
best_k = k_range[int(np.argmin(val_errors))]
print(f"Best value of k based on the validation error: {best_k}")

#Effect of rho on k
rho_values = [0.1, 0.3, 0.5]
plt.figure(figsize=(10, 6))
for r in rho_values:
    _, val_err_r = holdoutCVkNN(k_range, r, X_train, Y_train, num_rep=10)
    plt.plot(k_range, val_err_r, marker='o', label=f'Validation error (rho = {r})')
plt.xlabel('k (number of neighbours)')
plt.ylabel('Validation error')
plt.title('Effect of rho (fraction of points held out) on the choice of k')
plt.xticks(k_range)
plt.legend()
plt.grid(True)
plt.show()
print("Effect of rho and repetitions: with a small rho the validation set is small, so the validation error curve is noisy and the best k is unstable. Increasing rho gives a more reliable estimate of the error but leaves less data for training. A large number of repetitions averages out the randomness of a single hold-out split, giving a much more stable estimate of the best k.")

#Evaluate the performance on test set with the best hyper parameters ( i.e best k ).
knn_best = KNeighbourClassifier(n_neighbors=best_k)
knn_best.fit(X_train, Y_train)
test_predictions_best = knn_best.predict(X_test)
test_error_best = 1 - KNNAccuracy(Y_test, test_predictions_best)
print(f"Test error with the best k = {best_k}: {test_error_best:.4f} (accuracy = {1 - test_error_best:.4f})")

##################################
#Part(d): Influence of training data on KNN classifier
##################################
#Performance evaluation as n increases

#Function to generate the 4-Gaussian 2-class synthetic dataset (class 1: means (0,0), (1,1); class 2: means (0,1), (1,0))
def generate_synthetic_data(n_train_per_class, n_test_per_class, distribution="gaussian"):
    means_local = [[0, 0], [0, 1], [1, 0], [1, 1]]
    X_train_list, Y_train_list, X_test_list, Y_test_list = [], [], [], []
    for i, m in enumerate(means_local):
        label = 1 if i in [0, 3] else -1
        if distribution == "gaussian":
            samples = np.random.multivariate_normal(m, [[0.1, 0], [0, 0.1]], n_train_per_class + n_test_per_class)
        else:
            #Laplacian distribution: sample each dimension independently from a Laplace distribution
            #A Laplace distribution with scale b has variance 2*b^2, so b = sqrt(0.1/2) matches the Gaussian variance of 0.1
            b = np.sqrt(0.1 / 2)
            samples = np.stack((np.random.laplace(m[0], b, n_train_per_class + n_test_per_class),
                                np.random.laplace(m[1], b, n_train_per_class + n_test_per_class)), axis=1)
        X_train_list.append(samples[:n_train_per_class])
        X_test_list.append(samples[n_train_per_class:])
        Y_train_list += [label] * n_train_per_class
        Y_test_list += [label] * n_test_per_class
    return (np.vstack(X_train_list), np.array(Y_train_list),
            np.vstack(X_test_list), np.array(Y_test_list))

train_sizes = [200, 400, 1200]   #Total number of training points
k_range_d = list(range(1, 22, 2))
test_errors_n = []
for n in train_sizes:
    per_class = n // 4
    X_tr_n, Y_tr_n, X_te_n, Y_te_n = generate_synthetic_data(per_class, 2 * per_class)
    #Choose a good range for k as n changes using hold-out cross validation on the current training set
    _, val_err_n = holdoutCVkNN(k_range_d, 0.3, X_tr_n, Y_tr_n, num_rep=10)
    best_k_n = k_range_d[int(np.argmin(val_err_n))]
    #Evaluate the classifier (with the best k) on the test set
    knn_n = KNeighbourClassifier(n_neighbors=best_k_n)
    knn_n.fit(X_tr_n, Y_tr_n)
    test_error_n = 1 - KNNAccuracy(Y_te_n, knn_n.predict(X_te_n))
    test_errors_n.append(test_error_n)
    print(f"n = {n}: best k = {best_k_n}, test error = {test_error_n:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(train_sizes, test_errors_n, marker='o')
plt.xlabel('Training set size (n)')
plt.ylabel('Test error')
plt.title('KNN performance as the size of the training set grows')
plt.grid(True)
plt.show()

#Stability of the solution: check by repeating the validation multiple times
print("Stability of the best k (best k obtained in 10 independent hold-out CV runs):")
for n in train_sizes:
    per_class = n // 4
    X_tr_n, Y_tr_n, _, _ = generate_synthetic_data(per_class, 2 * per_class)
    best_ks = []
    for _ in range(10):
        _, val_err_r = holdoutCVkNN(k_range_d, 0.3, X_tr_n, Y_tr_n, num_rep=1)
        best_ks.append(k_range_d[int(np.argmin(val_err_r))])
    print(f"n = {n}: {best_ks}")
print("Observation: for small n the best k varies a lot between repetitions (unstable), while for large n the best k concentrates around a single value (stable). As n grows a larger k becomes optimal, since more training data reduces the estimation error of the neighbours.")

#Performance evaluation with different data distributions, i.e., Gaussian, Laplacian

#Generate both datasets with the same sizes as part (a): 120 training and 240 testing points
X_tr_g, Y_tr_g, X_te_g, Y_te_g = generate_synthetic_data(30, 60, distribution="gaussian")
X_tr_l, Y_tr_l, X_te_l, Y_te_l = generate_synthetic_data(30, 60, distribution="laplacian")

#Evaluate the KNN classifier with both l2 and l1 distance metrics on both datasets
for dist_name, (Xtr, Ytr_, Xte, Yte_) in [("Gaussian", (X_tr_g, Y_tr_g, X_te_g, Y_te_g)),
                                          ("Laplacian", (X_tr_l, Y_tr_l, X_te_l, Y_te_l))]:
    for metric in ["l2", "l1"]:
        knn_dist = KNeighbourClassifier(n_neighbors=5, distance_metric=metric)
        knn_dist.fit(Xtr, Ytr_)
        acc_dist = KNNAccuracy(Yte_, knn_dist.predict(Xte))
        print(f"{dist_name} data with {metric} metric: test accuracy = {acc_dist:.4f}")
print("Observation: the KNN achieves comparable performance on Gaussian and Laplacian distributed data, because both distributions separate the classes in a similar way. Changing the metric from l2 to l1 changes the results only slightly on this data; l1 is a bit more robust to the heavy tails of the Laplacian noise.")

##################################
#Part(e): Influence of distance metric on KNN classifier
##################################
#Performance evaluation of KNN classifier with different distance metric
metrics = ["l1", "l2", "linf"]
for metric in metrics:
    knn_e = KNeighbourClassifier(n_neighbors=5, distance_metric=metric)
    knn_e.fit(X_train, Y_train)
    acc_e = KNNAccuracy(Y_test, knn_e.predict(X_test))
    print(f"Accuracy with {metric} distance: {acc_e:.4f}")

#Understand the decision regions of KNN classifier with different distance metric
fig_e, axes_e = plt.subplots(1, 3, figsize=(18, 5))
x_min_e, x_max_e = X_train[:, 0].min() - 0.5, X_train[:, 0].max() + 0.5
y_min_e, y_max_e = X_train[:, 1].min() - 0.5, X_train[:, 1].max() + 0.5
xx_e, yy_e = np.meshgrid(np.linspace(x_min_e, x_max_e, 200), np.linspace(y_min_e, y_max_e, 200))
grid_points_e = np.c_[xx_e.ravel(), yy_e.ravel()]
for ax_e, metric in zip(axes_e, metrics):
    knn_e = KNeighbourClassifier(n_neighbors=5, distance_metric=metric)
    knn_e.fit(X_train, Y_train)
    Z_e = knn_e.predict(grid_points_e).reshape(xx_e.shape)
    ax_e.contourf(xx_e, yy_e, Z_e, alpha=0.3, cmap=ListedColormap(('blue', 'green')))
    ax_e.scatter(X_test[Y_test == 1][:, 0], X_test[Y_test == 1][:, 1], color='green', s=10, label='Class 1')
    ax_e.scatter(X_test[Y_test == -1][:, 0], X_test[Y_test == -1][:, 1], color='blue', s=10, label='Class 2')
    ax_e.set_title(f'Decision regions ({metric} metric)')
    ax_e.set_xlabel('X')
    ax_e.set_ylabel('Y')
    ax_e.legend()
plt.show()
print("Observation: the decision regions of all three metrics are similar for this well separated data, but the shapes of the boundaries differ locally: the l1 metric produces diamond-like (axis aligned) regions, the l2 metric produces circular regions and the l-infinity metric produces square regions. The differences are more visible in regions where the two classes overlap.")

##################################
#Part(f): MNIST Digit classification using KNN classifier
##################################

#Load MNIST data
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
mnist = fetch_openml('mnist_784')
images = mnist.data.to_numpy()
targets = mnist.target.to_numpy().astype(int)   #convert the string labels to integers
#Plot a few images
plt.subplot(211)
plt.imshow((images[0].reshape(28,28)), cmap=plt.cm.gray_r, interpolation='nearest')
plt.subplot(212)
plt.imshow(images[1].reshape(28,28), cmap=plt.cm.gray_r, interpolation='nearest')
plt.show()

#Create train, validation and test splits (80%, 10%, 10% of the entire dataset)
np.random.seed(0)
indices_mnist = np.random.permutation(len(images))
train_idx_m = indices_mnist[:int(0.8 * len(images))]
val_idx_m = indices_mnist[int(0.8 * len(images)):int(0.9 * len(images))]
test_idx_m = indices_mnist[int(0.9 * len(images)):]

X_train_mnist = images[train_idx_m].astype(np.float32)
Y_train_mnist = targets[train_idx_m]
X_val_mnist = images[val_idx_m].astype(np.float32)
Y_val_mnist = targets[val_idx_m]
X_test_mnist = images[test_idx_m].astype(np.float32)
Y_test_mnist = targets[test_idx_m]
print(f"Train: {X_train_mnist.shape}, Validation: {X_val_mnist.shape}, Test: {X_test_mnist.shape}")

#Write 'MultiClassKNNClassify' function
def MultiClassKNNClassify(X_train, Y_train, X_test, k=5, distance_metric="l2", chunk_size=500):
    #Inputs: training data, training labels, test data, number of neighbours, distance metric
    #Outputs: predicted labels for the test data using the majority vote among the k nearest neighbours
    #Distances are computed in a chunked, vectorized fashion to handle the large MNIST dataset
    X_train = np.asarray(X_train, dtype=np.float32)
    Y_train = np.asarray(Y_train)
    X_test = np.asarray(X_test, dtype=np.float32)
    predictions = []
    if distance_metric == "l2":
        train_norms = np.sum(X_train ** 2, axis=1)
    for start in range(0, len(X_test), chunk_size):
        X_chunk = X_test[start:start + chunk_size]
        if distance_metric == "l2":
            #||x - y||^2 = ||x||^2 + ||y||^2 - 2 x.y  (computed with a matrix product for efficiency)
            D = train_norms[None, :] + np.sum(X_chunk ** 2, axis=1)[:, None] - 2.0 * (X_chunk @ X_train.T)
        elif distance_metric == "l1":
            D = np.sum(np.abs(X_chunk[:, None, :] - X_train[None, :, :]), axis=2)
        elif distance_metric == "linf":
            D = np.max(np.abs(X_chunk[:, None, :] - X_train[None, :, :]), axis=2)
        else:
            raise ValueError("Unsupported distance metric: " + str(distance_metric))
        #k nearest neighbours of every test point in the current chunk
        nn_idx = np.argpartition(D, kth=min(k, D.shape[1] - 1), axis=1)[:, :k]
        neighbor_labels = Y_train[nn_idx]
        #Majority vote for every test point in the current chunk
        for row in neighbor_labels:
            values, counts = np.unique(row, return_counts=True)
            predictions.append(values[np.argmax(counts)])
    return np.array(predictions)

#Empirically chose most suitable k and distance metric based on the evauation on cross-validation data
#For computational tractability of the brute force KNN, the hyperparameter search is run on a random subsample of the data
sel_train_idx = np.random.permutation(len(X_train_mnist))[:4000]
sel_val_idx = np.random.permutation(len(X_val_mnist))[:1000]
X_tr_sel, Y_tr_sel = X_train_mnist[sel_train_idx], Y_train_mnist[sel_train_idx]
X_val_sel, Y_val_sel = X_val_mnist[sel_val_idx], Y_val_mnist[sel_val_idx]

candidate_ks = [1, 3, 5]
candidate_metrics = ["l1", "l2"]
best_val_acc, best_k_mnist, best_metric = -1, None, None
for metric in candidate_metrics:
    for k_cand in candidate_ks:
        chunk = 20 if metric == "l1" else 500
        preds_val = MultiClassKNNClassify(X_tr_sel, Y_tr_sel, X_val_sel, k=k_cand, distance_metric=metric, chunk_size=chunk)
        val_acc = np.mean(preds_val == Y_val_sel)
        print(f"k = {k_cand}, metric = {metric}: validation accuracy = {val_acc:.4f}")
        if val_acc > best_val_acc:
            best_val_acc, best_k_mnist, best_metric = val_acc, k_cand, metric
print(f"Best hyperparameters from cross-validation: k = {best_k_mnist}, metric = {best_metric} (validation accuracy = {best_val_acc:.4f})")

#Evaluate the performance on test data with the best hyper parameters ( k, distance metric ) obtained from cross validation
#Note: brute force l1 distances do not scale well with the training set size, so if l1 is selected a subsample of the training data is used
if best_metric == "l1":
    l1_train_idx = np.random.permutation(len(X_train_mnist))[:10000]
    X_tr_eval, Y_tr_eval = X_train_mnist[l1_train_idx], Y_train_mnist[l1_train_idx]
else:
    X_tr_eval, Y_tr_eval = X_train_mnist, Y_train_mnist

chunk_test = 10 if best_metric == "l1" else 500
test_preds_mnist = MultiClassKNNClassify(X_tr_eval, Y_tr_eval, X_test_mnist, k=best_k_mnist, distance_metric=best_metric, chunk_size=chunk_test)
test_acc_mnist = np.mean(test_preds_mnist == Y_test_mnist)
print(f"MNIST test accuracy with k = {best_k_mnist}, metric = {best_metric}: {test_acc_mnist:.4f}")

#Create a confusion matrix for test data
def compute_confusion_matrix(true, pred):
    #Inputs: Ground truth labels and classifier predictions
    #Outputs: Confusion matrix
    #Confusion matrix: A confusion matrix is a table that summarizes the performance of a classification model by comparing its predictions to the actual ground truth values
    #Rows: Represent the actual class labels
    #Columns: Represent the predicted class labels
    true = np.asarray(true)
    pred = np.asarray(pred)
    classes = np.unique(true)
    n_classes = len(classes)
    confusion_matrix = np.zeros((n_classes, n_classes), dtype=int)
    true_idx = np.searchsorted(classes, true)
    pred_idx = np.searchsorted(classes, pred)
    np.add.at(confusion_matrix, (true_idx, pred_idx), 1)
    return confusion_matrix, classes

conf_matrix, classes_mnist = compute_confusion_matrix(Y_test_mnist, test_preds_mnist)

#Visualize the confusion matrix
plt.figure(figsize=(8, 7))
plt.imshow(conf_matrix, cmap='Blues')
plt.colorbar()
plt.xticks(range(len(classes_mnist)), classes_mnist)
plt.yticks(range(len(classes_mnist)), classes_mnist)
plt.xlabel('Predicted label')
plt.ylabel('True label')
plt.title('Confusion matrix (MNIST test data)')
for i in range(len(classes_mnist)):
    for j in range(len(classes_mnist)):
        plt.text(j, i, str(conf_matrix[i, j]), ha='center', va='center',
                 color='red' if (i != j and conf_matrix[i, j] > 0) else 'black', fontsize=8)
plt.show()

#Most confused classes (digits): largest off-diagonal counts
conf_offdiag = conf_matrix.copy()
np.fill_diagonal(conf_offdiag, 0)
print("Most confused digit pairs (true -> predicted):")
flat_idx = np.argsort(conf_offdiag.ravel())[::-1][:5]
for idx_f in flat_idx:
    i_f, j_f = np.unravel_index(idx_f, conf_offdiag.shape)
    if conf_offdiag[i_f, j_f] > 0:
        print(f"  true {classes_mnist[i_f]} -> predicted {classes_mnist[j_f]}: {conf_offdiag[i_f, j_f]} times")

#Suggest an alternative ways to improve performance
print("Ways to improve the KNN performance on MNIST:")
print("1. Use dimensionality reduction (e.g., PCA) before KNN to remove redundant pixels and speed up the distance computations.")
print("2. Use data augmentation (small translations/rotations of the digits) to make the classifier invariant to small shifts.")
print("3. Use distance-weighted voting (weight the neighbours by 1/distance) instead of simple majority voting.")
print("4. Use approximate nearest neighbour search structures (e.g., KD-trees, LSH) to reduce the prediction time.")
print("5. Use feature engineering (blur/edge descriptors) or switch to more powerful models (SVM, neural networks).")
```
